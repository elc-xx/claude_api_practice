```python
# Load env variables
from dotenv import load_dotenv

load_dotenv()

# Create an API client
import os
from anthropic import Anthropic

client = Anthropic()
model = os.getenv("CLAUDE_MODEL", "")
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)
```

```python
# Basic Implementation of a streaming response

messages = []
add_user_message(messages, "Write a 1 sentence description of a fake music database")

stream = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages,
    stream=True
)

# Print the RAW events as they come in
for event in stream:
    print(event)

```

```python
# Simplified Text Streaming
# This approach automatically filters out everything except the actual text content, which is usually what you need for displaying responses to users.

messages = []
add_user_message(messages, "Write a 1 sentence description of a fake music database")

with client.messages.stream(
    model=model,
    max_tokens=1000,
    messages=messages
) as stream:
    for text in stream.text_stream:
        # Send each chunk to your client
        print(text, end="")
    
    # Get the complete message for database storage
    final_message = stream.get_final_message()
```
